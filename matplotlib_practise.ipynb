```python
import numpy as np 
import matplotlib.pyplot as plt 

year = np.array([2023, 2024, 2025, 2026])
sale = np.array([15,25, 30,20])
sale_2 = np.array([5,25,10,1])
plt.plot(year,sale, marker='o',
                    markersize=30,
                    markerfacecolor='cyan',
                    markeredgecolor='cyan',
          color = 'Red',
          linestyle='solid',
          linewidth= 4
                    )
plt.plot(year,sale_2, marker='o',
                    markersize=15,
                    markerfacecolor='blue',
                    markeredgecolor='cyan',
          color = 'Red',
          linestyle='solid',
          linewidth= 4
                    )
plt.show()
```

```python
# excercise2
year = np.array([2023, 2024, 2025, 2026])
sale = np.array([15,25, 30,20])
sale_2 = np.array([5,25,10,1])

plt.plot(year, sale)
plt.plot(year,sale_2)
plt.title("Practical",fontsize=19,fontweight='bold',color='blue', family='Arial')
plt.xlabel('Years')
plt.ylabel('Sales')
plt.xticks(year)
plt.tick_params(axis='both', colors='green')
plt.show()
```

```python
# Excercise4
# bar chart() = use to compare categories of data by representing each category wih a bar
import numpy as np 
import matplotlib.pyplot as plt 
categories = np.array(['Grains','fruit','vegetables','Protein','Dairy','Sweets'])
values= np.array([4,3,2,5,3,1])

plt.barh(categories,values, color= 'green')
plt.title('Daily Consumption')
plt.xlabel('Foods')
plt.ylabel('Quantities')
plt.show()
```

```python
# Excercise5
# Pie chart = Circular chart divided into slices to show percentages of the total.
#             Good for visualizing disribution amount categories.

categories = np.array(['Freshmen','Sophomores','Juniors','Senior'])
values = np.array([300, 250, 275, 225])
colors = ['red','yellow','blue','green']

plt.pie(values,labels=categories,autopct='%1.1f',colors=colors,explode=[0,0,0,0.2],shadow=True,startangle=90)
plt.title('ACLEDA UNIVERSITY OF BUSINESS', color='green',fontweight='bold',fontsize=19)
plt.show()
```

```python
# scatter graph = Shows the rlationship between two variables
#                 Helps to identify a correlation (+, -, None)
#                 Example: Study hours vs Test scores

x = np.array([0,1,1,2,3,4,5,6,7,7,8]) # study hours
y = np.array([55,60,65,62,68,70,75,78,82,85,87]) # score test

x2 = np.array([0,1,2,2,3,4,5,6,6,8,8]) # study hours
y2 = np.array([50,55,65,62,66,70,75,78,80,84,90]) # score test

plt.scatter(x,y,color='blue',s= 100, alpha=0.5,label='Class A')
plt.scatter(x2,y2,color='red',s = 70, alpha=0.5, label='Class B')

plt.legend()
plt.show()
```

```python
# Figure = The entire canvas 
# Ax = A single plot (subplot)
# suplots(rows, columns)


x = np.array([1,2,3,4])
figure, axes = plt.subplots(2,2)

axes[0,0].plot(x, x*2,color='red')
axes[0,0].set_title('X* 2')

axes[0,1].bar(x, x**2, color='blue')
axes[0,1].set_title('X** 2')

axes[1,0].plot(x, x**3, color='green')
axes[1,0].set_title('X** 3')

axes[1,1].plot(x, x**4, color='yellow')
axes[1,1].set_title('X** 4')


plt.show()
```
